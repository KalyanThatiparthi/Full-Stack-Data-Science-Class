# Function with Return Function

```python
def wish():
    print('Hello, Welcome')
wish()
wish()
```

```python
def wish():
    print('Hello')
def wish():
    print('Good Morning')
wish()
```

```python
def wish():
    print('Hello')
def greet():
    print('Good Morning')
wish()
greet()
```

```python
def wish():
    return 'Hello, Good Morning!!'
res=wish()
print(res)
```

# Function with Arguments and Without Arguments

```python
def add():
    a,b=5,3
    print(a+b)
add()
```

```python
def div(a,b):
    return a//b
res=div(6,3)
print(res)
```

```python
def sub(a,b):
    c=a-b
    print(c)
sub(6,3)
```

```python
def multi(a,b):
    return a*b
res=multi(4,3)
print(res)
```
