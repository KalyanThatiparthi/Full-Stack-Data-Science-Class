# Array Manipulation Funtions

```python
a1=np.array([[1,2,3],[4,5,6]])
print('Actual Array: \n',a1) 
reshaped=np.reshape(a1,(3,2))
print()
print('Reshaped Array: \n',reshaped)
```

```python
# Flattened Array
flatten=np.ravel(reshaped)
print('Flattend Array: \n', flatten)
```

```python
# Transpose an Array
a2=np.array([[1,2],[3,4],[5,6]])
print('Array: \n', a2)
tr=np.transpose(a2)
print()
print('Transposed Array: \n', tr)
```

```python
#Stack Array Vertically
a3=np.array([1,2,3])
a4=np.array([4,5,6])
a5=np.array([7,8,9])
vstack=np.vstack([a3,a4,a5])
print('Vertical Stack Array: \n', vstack)
```

```python
#Stack Array Horizantally
a3=np.array([1,2,3])
a4=np.array([4,5,6])
a5=np.array([7,8,9])
hstack=np.hstack([a3,a4,a5])
print('Horizantal Stack Array: \n', hstack)
```

# SET Operations

```python
set1=np.array([1,2,3,4,5])
set2=np.array([4,5,6,7,8])
print('2 Sets:', '\nSET 1:',set1, '\nSET 2:',set2)
```

```python
np.intersect1d(set1,set2)
```

```python
np.union1d(set1,set2)
```

```python
np.unique_counts((set1,set2))
```

```python
np.unique_all((set1,set2))
```

```python
np.unique_inverse((set1,set2))
```

```python
arr = np.array([1, 2, 3, 4, 5])
shape = arr.shape
size = arr.size
dimensions = arr.ndim # Number of dimensions
dtype = arr.dtype # Data type of the array elements
print("Shape of a: ", shape)
print("Size of a: ", size)
print("Number of dimensions of a: ", dimensions)
print("Data type of a: ", dtype)
type_change=np.astype(arr,float)
print('Changing Data Type: ',type_change)
```
